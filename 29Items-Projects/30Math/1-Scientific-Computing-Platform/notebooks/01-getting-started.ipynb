# Scientific Computing Platform — Getting Started

This notebook demonstrates the `sciengine` kernel that powers the whole platform:
the **same library** runs behind the web API, the compute workers, and these notebooks.

Local setup (from the repo root): `uv sync` — then select the project venv as the kernel.
On AWS, run this in SageMaker Studio with the platform kernel image (Phase 2).

```python
from sciengine.symbolic.solver import solve_equation

result = solve_equation("x^2 - 4 = 0", variable="x")
print("solutions:", result.solutions)
print("as LaTeX :", result.solutions_latex)
```

```python
from sciengine.numerical.roots import bisect

root = bisect(lambda x: x**2 - 2.0, 0.0, 2.0)
print(f"sqrt(2) \u2248 {root.root:.12f} after {root.iterations} bisection steps")
```

```python
from IPython.display import SVG

from sciengine.plotting.function_plot import plot_sympy_expression

SVG(plot_sympy_expression("sin(x) + sin(3x)/3", "x", -6.5, 6.5))
```
